```python
#indirect recursion
#code to check the given numbers is even or odd using In-Direct Recursion
def even(n):
  if n==0:
    print("Even")
    return
  odd(n-1)
def odd(n):
  if n==0:
    print("Odd")
    return
  even(n-1)
n=int(input("enter a number: "))
even(n)
```

```python
#Tree Recursion-fibonacci number
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
n=int(input("Enter a number: "))
print("Fibonacci:",fib(n))
```

```python
# spiral order pattern
n=int(input("enter the size:"))
a=[[0] * n for _ in range(n)]
print(a)
top=0
bottom=n-1
left = 0
right =n-1
num=1
while top<=bottom and left<=right:
  # traverse at top from left-> right
  for i in range(left, right+1):
    a[top][i] = num
    num+=1
  top+=1
  # traverse at right from top ->bottom
  for i in range(top,bottom+1):
    a[i][right] = num
    num+=1
  right -=1
  # traverse at bottom from right-> left
  for i in range(right,left-1,-1):
    a[bottom][i]=num
    num+=1
  bottom-=1
  # traverse at left from bottom-> top
  for i in range(bottom,top-1,-1):
    a[i][left]=num
    num+=1
  left+=1
for row in a:
  for val in row:
    print(f"{val:3}", end=" ")
  print()


```

```python
# sorting searches
# 1.bubble sort
arr=list(map(int,input("enter elements:").split()))
n=len(arr)
for i in range(n):
  for j in range(0,n-1-i):
    if arr[j]>arr[j+1]:
      arr[j] , arr[j+1]=arr[j+1] , arr[j]
print(*arr)
```

```python
#selection sort
arr=list(map(int,input("enter elements:").split()))
n=len(arr)
for i in range(n-1):
  minindex=i
  for j in range(i+1,n):
    if arr[j]<arr[minindex]:
      minindex=j
  arr[i],arr[minindex]=arr[minindex],arr[i]
print(*arr)
```

```python
# insertion sort
arr=list(map(int,input("enter elements:").split()))
for i in range(1,len(arr)):
  key=arr[i]
  j=i-1
  while j>=0 and arr[j]>key:
    arr[j+1] = arr[j]
    j-=1
  arr[j+1] = key
print(*arr)
```

```python
# divide and conquer ---- mid partition
# 1. merge sort
#Divide and conquer --------------------Mid partition
#1.Merge sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    result = []
    i,j=0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
arr = list(map(int, input("Enter elements:").split()))
print("original array", arr)
sorted_array=merge_sort(arr)
print("sorted array", *sorted_array)
```

```python

#quick sort
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1
def quick_sort(arr, low, high):
    if low < high:
        p = partition(arr, low, high)
        quick_sort(arr, low, p - 1)
        quick_sort(arr, p + 1, high)
arr = list(map(int, input("Enter elements: ").split()))
print("Original array",arr)
quick_sort(arr, 0, len(arr) - 1)
print("Sorted array:", arr)


```

```python
# COMPETITIVE CODING
# PANCAKE SORT /FLIP SORT
def flip(arr,k):
  arr[:k+1]=arr[:k+1][::-1]
def pancakesort(arr):
  n=len(arr)
  for size in range(n,1,-1):
    max_index=arr.index(max(arr[:size]))
    if max_index !=size-1:
      if max_index!=0:
        flip(arr,max_index)
      flip(arr,size-1)
n=int(input("enter array size:"))
arr=list(map(int,input("enter elements").split()))
pancakesort(arr)
print("sorted array:",*arr)
```
