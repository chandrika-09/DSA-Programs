```python
# write a code to print the sum of digits of a given number
n=int(input("enter a number :"))
sum=0
while n!=0:
  d=n%10
  sum+=d
  n//=10
print("sum of digits", sum)
```

```python
#write a code to print the reverse of a given number.
n=int(input("Enter a number:"))
rev=0
while n>0:
    d=n%10
    rev=rev*10+d
    n//=10
print("reverse number:",rev)
```

```python
# write a code to print the count of even digits and odd  digits of a given number
n=int(input("enter a number"))
even=0
odd=0
while n!=0:
  d=n%10
  if d%2==0:
    even+=1
  else:
    odd+=1
  n //=10
print("even count",even)
print("odd count",odd)

```

**RECURSIONS**

```python
#Direct Recursion
#code to print n natural numbers using direct recursion
def numbers(n):
  if n==0:
    return
  print(n,end=' ')
  numbers(n-1)
n=int(input("Enter a numbers: "))
numbers(n)
```

```python
#indirect recursion
#code to check the given numbers is even or odd using In-Direct Recursion
def even(n):
  if n==0:
    print("Even")
    return
  odd(n-1)
def odd(n):
  if n==0:
    print("Odd")
    return
  even(n-1)
n=int(input("enter a number: "))
even(n)
```

```python
#Tree Recursion-fibonacci number
def fib(n):
  if n<=1:
    return n
  return fib(n-1)+fib(n-2)
n=int(input("Enter a number: "))
print("Fibonacci:",fib(n))
```

```python
# math approach
num=int(input("enter a value"))
if num%2==1:
  print(num,  "is odd")
else:
  print(num, "is even")
```

```python
# naive approach
arr=list(map(int,input("enter elements").split()))
largest = arr[0]
for i in range(len(arr)):
  if arr[i]>largest:
    largest=arr[i]
print("largest:", largest)
```

```python

##Brute-Force approach
arr=list(map(int,input("Enter elements").split()))
target=int(input("Enter target:"))
found=False
for i in range(len(arr)):
  for j in range(i+1,len(arr)):
    if arr[i]+arr[j]==target:
      print("Pair Found at:",i,j)
      print("Pair Found are:",arr[i],arr[j])
      found=True
      break
  if found:
    break
if not found:
  print("No pair found")
```

```python
# greedy approach
coins=list(map(int,input("enter elements").split()))
amount = int(input("enter target:"))
print("coins used:")
for coin in coins:
  while amount>=coin:
    print(coin,end=" ")
    amount -= coin
```

```python
# back tracking
def generate(s,n):
  if len(s)==n:
    print(s)
    return
  generate(s+'0',n)
  generate(s+'1', n)
n=int(input("enter length:"))
generate("",n)
```
